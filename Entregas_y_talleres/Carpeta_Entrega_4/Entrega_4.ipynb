#### ${\color{#c7c7c7}{{\textbf{{{Ejercicio 1}}}}}}$

```python
#Ejercicio 1 (funciones):
import math 
#Definición función
def area_circu(r):
    """
    Esta función calcula el área de un círculo de radio r dado.
    
    Parámetros
    r (float): Radio del círculo
    
    Retorno:
    A, o sea, el área del círculo
    """
    return (math.pi)*r**2

print(area_circu.__doc__)
print("----------------------------\n")
print("Áreas de tres círculos con radios 1,2 y 3 respectivamente:","\n")
print(area_circu(1),"\n")
print(area_circu(2),"\n")
print(area_circu(3))
```

#### ${\color{#913789}{{\textbf{{{Ejercicio 3}}}}}}$

```python
#ejercicio 3 (funciones):
#definición función
def prom(a,b,c):
    """
    Esta función calcula el promedio de los tres números
    
    Parámetros
    a(float): Número cualquiera dado
    b(float): Número cualquiera dado
    c(float): Número cualquiera dado
    
    Retorno
    Devuelve el promedio de los tres números introducidos
    
    """
    return((a+b+c)/3)

#Entrada y salida
print(prom.__doc__)
print("----------------------------\n")
a=float(input("Introduzca el primer número: ",))
b=float(input("Introduzca el segundo número: ",))
c=float(input("Introduzca el tercer número: ",))
print("El promedio de los números fue de: ",prom(a,b,c))
```

```python
#Ejercicio 4 (funciones):
#definición función
def es_par(n):
    """
    Esta función verifica si un número es par o no. Devuelve un valor booleano en cualquier caso
    
    Parámetros
    n(int): Número cualquiera a verificar
    
    Retorno
    Devuelve un valor booleano tras verificar la condición, True si es par y False si es impar.
    """
    return(n%2==0)

#Entrada y salida
print(es_par.__doc__)
print("----------------------------\n")
n=int(input("Introduzca el número a verificar su paridad:",))
if(es_par(n)==True):
    print("El número es par")
else:
    print("El número es impar")
```

#### ${\color{#861815}{{\textbf{{{Ejercicio 5}}}}}}$

```python
#Ejercicio 5 (funciones):
#definición funciones
def celsius_fahrenheit(c):
    """
    Convierte una temperatura en grados celsius a grados fahrenheit
    
    Parámetros
    c(float): Temperatura dada en grados celsius
    
    Retorno
    Devuelve una temperatura en grados fahrenheit
    """
    return (9/5)*c + 32

def fahrenheit_celsius(f):
    """
    Convierte una temperatura de grados fahrenheit a grados celsius
    
    Parámetros
    f(float): Temperatura dada en grados fahrenheit
    
    Retorno
    Devuelve una temperatura en grados celsius
    """
    return (f-32)*(5/9)

print(celsius_fahrenheit.__doc__)
print("----------------------------\n")
print(fahrenheit_celsius.__doc__)
print("----------------------------\n")
c=float(input("Ingrese una temperatura en grados celsius:",))
print("Su temperatura convertida a grados fahrenheit es de:",celsius_fahrenheit(c))
print("Su temperatura inicial fue de:",fahrenheit_celsius(celsius_fahrenheit(c)))
```

#### ${\color{#64b8f5}{{\textbf{{{Ejercicio 6}}}}}}$

```python
#ejercicio 6 (funciones):
#definición funciones
def distancia(v,t):
    """
    Calcula una distancia d a partir de dos valores iniciales de velocidad y tiempo
    
    Parámetros
    v(float): Velocidad inicial constante dada de un objeto
    t(float): Tiempo hasta el cual se quiere determinar la distancia que recorre un objeto
    
    Retorno
    Devuelve la distancia total recorrida por un objeto en MRU
    """
    return (v*t)

def pos_f(x0,x):
    """ 
    Calcula la posición final de un objeto con base en unos valores de posición inicial y distancia recorrida
    
    Parámetros 
    x0(float): Posición inicial dada
    x(float): Distancia recorrida en un intervalo de tiempo t
    
    Retorno
    Devuelve la posición final de un objeto 
    """
    return(x0+x)

print(distancia.__doc__)
print("----------------------------\n")
print(pos_f.__doc__)
print("----------------------------\n")
x0=float(input("Ingrese la posición inicial del objeto:",))
v=float(input("Ingrese la velocidad inicial del objeto:",))
t=float(input("Ingrese el tiempo hasta el cual el objeto se mueve:",))
print("La distancia total recorrida por el objeto en un tiempo t fue de:",distancia(v,t))
print("La posición final del objeto según los datos es de:",pos_f(x0,distancia(v,t)))
```

#### ${\color{#505050}{{\textbf{{{Ejercicio 8}}}}}}$

```python
#ejercicio 8 (funciones):
#definición funciones
def E_p(m,h,g=9.81):
    """ 
    Calcula la energía potencial de un objeto respecto a una altura
    
    Parámetros
    m(float): Masa del objeto 
    g(float): Valor de la gravedad en la tierra
    h(float): Altura respecto a un punto del objeto
    
    Retorno
    Devuelve el valor de la energía potencial asociada al objeto
    """
    return(m*g*h)

def E_c(m,v):
    """ 
    Calcula la energía cinética de un objeto
    
    Parámetros
    m(float): Masa del objeto 
    v(float): Velocidad del objeto
    
    Retorno 
    Devuelve el valor de la energía cinética asociada al objeto
    """
    return (0.5*m*(v**2))

def E_m(m,v,h,g=9.81):
    """ 
    Calcula la energía mecánica de un objeto
    
    Parámetros
    m(float): Masa del objeto 
    v(float): Velocidad del objeto
    g(float): Valor de la gravedad en la tierra
    h(float): Altura respecto a un punto del objeto
    
    Retorno 
    Devuelve el valor de la energía mecánica asociada al objeto
    """
    return(E_p(m,h,g) + E_c(m,v))

print(E_m.__doc__)
print("----------------------------\n")
m=float(input("Ingrese la masa del objeto:",))
v=float(input("Ingrese la velocidad del objeto:",))
h=float(input("Ingrese la altura respecto a un punto del objeto:",))
print("La energía mecánica asociada al objeto fue de: ",E_m(m,v,h))
```

```python
#Ejercicio 7 -- Reto integrador

totales=[]
nombres=[]
precios=[]
cantidades=[]

def agregar_y_total(nombre,precio,cantidad):
        nombres.append(nombre)
        precios.append(precio)
        cantidades.append(cantidad)
        totales.append(cantidad*precio)


try:
        nombre=input("Ingrese el nombre del producto: ",)
        precio=float(input("Ingrese el precio del producto: ",))
        if precio<0:    
                raise ValueError ("Precio negativo, vuelva a ingresarlo...")
        cantidad=int(input("Ingrese la cantidad comprada del producto: ",))
        if cantidad<0:
                raise ValueError ("Cantidad negativa, vuelva a ingresarla...")
        agregar_y_total(nombre,precio,cantidad)
except ValueError as error:
        print("Error de valor, vuelva a intentarlo: ",error)
else:
        print("Este es el resumen:\n")
        print("- Nombre: ",nombre,"\n")
        print("- Precio: ",precio,"\n")
        print("- Cantidad: ",cantidad,"\n")
        print("- Total: ",precio*cantidad,"\n")
finally:
        print("Código finalizado...")
```
