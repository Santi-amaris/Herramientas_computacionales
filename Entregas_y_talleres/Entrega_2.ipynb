#### $\color{Plum} {\underline{\textbf{Ejercicio 2 -- Analizador de una calificación}}}$

```python
#Entrada de datos
nombre_estudiante = input("Ingrese el nombre del estudiante: ", )
nota_examen_1 = round(float(input("Ingrese el valor de la primera nota (entre 0.0 y 5.0): ", )), 1)
nota_examen_2 = round(float(input("Ingrese el valor de la segunda nota (entre 0.0 y 5.0): ", )), 1)
nota_proyecto_final = round(float(input("Ingrese el valor de la nota del proyecto final (entre 0.0 y 5.0): ", )), 1)
promedio= (nota_examen_1 + nota_examen_2 + nota_proyecto_final)/3

#Procesamiento de datos (aritméticos)
nota_max= max(nota_proyecto_final, nota_examen_2, nota_examen_1)
nota_min= min(nota_proyecto_final, nota_examen_2, nota_examen_1)
dif_min_max= nota_max - nota_min
#Procesamiento de datos (lógicos)
nota_mayor_3 = (nota_examen_1>=3.0) or (nota_examen_2>=3.0) or (nota_proyecto_final>=3.0)
notas_mayor_4 = (((nota_examen_1) and (nota_examen_2)) >= 4.0) or (((nota_examen_1) and (nota_proyecto_final)) >= 4.0) or (((nota_examen_2) and (nota_proyecto_final)) >= 4.0)
notas_mayores_3 = ((nota_examen_1 and nota_examen_2 and nota_proyecto_final) >=3)
nota_igual_5= (nota_examen_1 == 5.0) or (nota_examen_2 == 5.0) or (nota_proyecto_final == 5.0)

#Salida de datos
print("Este es el informe de notas del/la estudiante [", nombre_estudiante, "]")
print("--------------------------------------")
print("El promedio final fue de: ", round(promedio,1))
print("--------------------------------------")
print("La nota máxima obtenida fue de: ", nota_max)
print("--------------------------------------")
print("La nota mínima obtenida fue de: ", nota_min)
print("--------------------------------------")
print("La diferencia entre la nota máxima y mínima obtenida fue de: ", dif_min_max)
print("--------------------------------------")
print("¿Obtuvo al menos una nota superior o igual a 3.0?: ", nota_mayor_3)
print("--------------------------------------")
print("¿Obtuvo al menos dos notas iguales o superiores a 4.0?: ", notas_mayor_4)
print("--------------------------------------")
print("¿Todas las notas son superiores o iguales a 3.0?: ", notas_mayores_3)
print("--------------------------------------")
print("¿Obtuvo algún 5.0?: ", nota_igual_5)
```

#### $\color{BurlyWood}{\underline{\textbf{Ejercicio 4 -- Conversor de unidades de tiempo}}}$

```python
#Entrada de datos
cant_seg=int(input("Ingrese la cantidad de segundos deseada:",))

#Procesamiento de datos [lógicos]
cant_sup_1hora= cant_seg > 3600
cant_sup_1dia= cant_seg > 86400
cant_div_60= (cant_seg%60)==0
#Procesamiento de datos [aritméticos]
conver_hora= divmod(cant_seg,3600)
conver_min= divmod(conver_hora[1],60) #la conversión de minutos tiene como entrada el residuo de la anterior operación para evitar duplicidad
cant_tot_min= divmod(cant_seg,60)
seg_recons= (conver_hora[0]*3600) + (conver_min[0]*60) + conver_min[1]

#Salida de datos
print("Su cantidad de segundos convertida a horas, minutos, segundos es: ")
print("--------------------------------------")
print("Horas:", conver_hora[0])
print("--------------------------------------")
print("Minutos:", conver_min[0])
print("--------------------------------------")
print("Segundos:", conver_min[1])
print("--------------------------------------")
print("La cantidad total de minutos es de:", cant_tot_min[0])
print("--------------------------------------")
print("La cantidad total de horas es de:", conver_hora[0])
print("--------------------------------------")
print("La diferencia entre la cantidad de segundos introducida y la reconstruida es de:", (cant_seg - seg_recons))
print("--------------------------------------")
print("¿La cantidad de segundos supera la hora?:", cant_sup_1hora)
print("--------------------------------------")
print("¿La cantidad de segundos supera un día?:", cant_sup_1dia)
print("--------------------------------------")
print("¿La cantidad de segundos es divisible entre 60?:", cant_div_60)
```

#### $\color{LightSalmon}{\underline{\textbf{Ejercicio 6 -- Asesor de viajes}}}$

```python
#Entrada de datos
nombre_viajero=input("Ingrese el nombre de la persona encargada:", )
segundos_nom_viajero= int(input("Ingrese la cantidad de segundos nombres que tiene:", ))
apellidos_viajero=int(input("Ingrese la cantidad de apellidos que tiene:", ))
edad_viajero= int(input("Ingrese la edad de la persona encargada:", ))
ciudad_origen= input("Ingrese la ciudad desde la que partirá el vuelo:", )
ciudad_destino= input("Ingrese la ciudad desde a la que llegará el vuelo:", )
precio_vuelo_per= float(input("Ingrese el precio de vuelo por persona:", ))
precio_alojamiento_per= float(input("Ingrese el precio de alojamiento por noche:", ))
num_noches= int(input("Ingrese la cantidad de noches de hospedaje:", ))
presupuesto= float(input("Ingrese su presupuesto disponible:", ))
num_personas= int(input("Ingrese la cantidad de personas que viajarán:", ))

#Procesamiento de datos [aritméticos]
costo_aloj= (precio_alojamiento_per*num_noches*num_personas)
costo_tot= ((precio_vuelo_per*num_personas) + costo_aloj)
dinero_res= presupuesto - costo_tot
cost_prom_per= (costo_tot/num_personas)
cost_prom_noche= (costo_aloj/num_noches)
canti_carac=(len(nombre_viajero) - (segundos_nom_viajero+apellidos_viajero))
dif_abs= abs(presupuesto - costo_tot)
#Procesamiento de datos [lógicos]
presup_sufi= presupuesto>costo_tot
precio_sup_5m= costo_tot>5000000
cant_pers_sup_2= num_personas>=2
viajero_mayor= edad_viajero>=18
viaje_dura_sem= num_noches>=8
viaje_econoc= (costo_tot<=3000000) and (num_noches<=7)

#Salida de datos
print("Bienvenido [",nombre_viajero,"], este es el análisis de su viaje: ")
print("--------------------------------------")
print("El costo total de hospedaje es de:", costo_aloj)
print("--------------------------------------")
print("El costo total del viaje, incluyendo vuelo y alojamiento pero sin incluir gastos varios, es de:", costo_tot)
print("--------------------------------------")
print("Su dinero restante tras los gastos calculados es de:", dinero_res)
print("--------------------------------------")
print("El costo promedio por persona es de:", cost_prom_per)
print("--------------------------------------")
print("El costo promedio por noche es de:", cost_prom_noche)
print("--------------------------------------")
print("La cantidad de caracteres del nombre de la persona encargada es de:", canti_carac)
print("--------------------------------------")
print("La diferencia absoluta entre el presupuesto y el costo total del viaje fue de:", dif_abs)
print("--------------------------------------")
print("¿El presupuesto es suficiente para el viaje?:", presup_sufi)
print("--------------------------------------")
print("¿El precio del viaje supera los $5000000?:", precio_sup_5m)
print("--------------------------------------")
print("¿Viajarán dos o más personas?:", cant_pers_sup_2)
print("--------------------------------------")
print("¿La persona encargada es mayor de edad?:", viajero_mayor)
print("--------------------------------------")
print("¿El viaje dura más de una semana?:", viaje_dura_sem)
print("--------------------------------------")
print("¿El viaje es económico?:", viaje_econoc)
```
