#### $\hspace{6cm}\color{#d80059}{{\textbf{Ejercicio 1 — Tupla de coordenadas}}}$

```python
#Definición, desempaquetado
punto=(8,12)
x,y=punto
d=0

#Procesamiento
d= abs(0-x) + abs(0-y)

#Salida
print("La coordenada x es:",x)
print("La coordenada y es:",y)
print('La distancia al "origen" de Manhattan (punto) es:',d)
```

#### $\hspace{6cm}\color{#f4a3b3}{{\textbf{Ejercicio 3 — Valores únicos}}}$

```python
#Entrada
numeros = [4, 7, 4, 2, 9, 7, 2, 2, 10, 9]
numeros_conjuntos=set()
#Conversión lista-conjunto
for num in numeros:
    numeros_conjuntos.add(num)
conjun_10={10}
conjun_5={5}

#Procesamiento y salida

print("En el conjunto definido existen",len(numeros_conjuntos),"valores diferentes") #Por la particularidad de la unicidad de elementos en los conjuntos...

if((conjun_10<=numeros_conjuntos)==True):
    print('El número 10 pertenece al conjunto "números"')
else:
    print('El número 10 no pertenece al conjunto "números"')

if((conjun_5<=numeros_conjuntos)==True):
    print('El número 5 pertenece al conjunto "números"')
else:
    print('El número 5 no pertenece al conjunto "números"')
```

#### $\hspace{6cm}\color{#4900b4}{{\textbf{Ejercicio 4 — Comparación de grupos}}}$

```python
#Entrada
programacion = {"Ana", "Luis", "María", "Carlos", "Sofía"}
robotica = {"Luis", "Carlos", "Sofía", "Juan", "Pedro"}

#Procesamiento y salida
print("Los estudiantes que participan en ambas son:",programacion&robotica)
print("Los estudiantes que participan solamente en programación son:",programacion-robotica)
print("Los estudiantes que participan solamente en robótica son:",robotica-programacion)
print("Los estudiantes que participan en al menos una actividad son:",robotica|programacion)
```

#### $\hspace{6cm}\color{#25c725}{{\textbf{Ejercicio 5 — Diccionario de productos}}}$

```python
#Entrada, definición
diccionario_produc={
    "Bloqueador solar":5600,
    "Agua con gas":2700,
    "Detergente":8900,
    "Cable de cobre":15000,
    "Audifonos":68900,
    "Jamón":7000,
    "Cuajada":11000
}
#Procesamiento de datos y salida
consulta=(input("¿De cuál producto desea consultar el precio? Ingréselo a continuación:",)).capitalize()

print("La lista de productos es:")
for i in diccionario_produc.keys():
    print("-",i)
print()

print("La lista de precios (respectivamente) es:")
for i in diccionario_produc.values():
    print("-",i)
print()

print("La lista de productos con sus respectivos precios es:")
for clave,valor in diccionario_produc.items():
    print("Producto:",clave, "| ","Precio:",valor)
print()

if((diccionario_produc.get(consulta))!=None):
    print("El precio del producto [",consulta,"] es:",(diccionario_produc.get(consulta)))
else:
    print("El producto que consultó no está registrado")
```

#### $\hspace{6cm}\color{#e2b847}{{\textbf{Ejercicio 6 — Notas de estudiantes}}}$

```python
#Entrada y definición de datos
registro_estudiantes={
    "Pedro":2.9,
    "Álvaro":4.1,
    "Diana":3.7,
    "Gertrudis":4.8,
    "Antonia":1.5
}
verif_aprob=""
verif_aprob_conjun={}
cont_aprob=0
cont_noaprob=0

#Procesamiento y salida
for i in registro_estudiantes.values():
    if (i>=3.0):
        verif_aprob="Aprobó"
        cont_aprob+=1
    else:
        verif_aprob="No aprobó"
        cont_noaprob+=1
    verif_aprob_conjun[i]=verif_aprob

for clave,i in registro_estudiantes.items():
    print(clave,":",i,"-",verif_aprob_conjun[i])
print()
print("Aprobaron [",cont_aprob,"] estudiantes")
print("No aprobaron [",cont_noaprob,"] estudiantes")
```

#### $\hspace{6cm}\color{SkyBlue}{{\textbf{Ejercicio 7 — ¿Qué estructura elegirías?}}}$
* Caso 1: Para este caso eligiría una tupla, ya que si nos dan una coordenada (punto fijo en el espacio), siempre y cuando no se den más datos iniciales, yo asumiría que esta no cambia y la estructura de tupla ayuda a preservarla (por defición la tupla no es mutable).

* Caso 2: En esta ocasión yo eligiría un conjunto para aprovechar la particularidad de que en los conjuntos todos los elementos son únicos por más veces que se repitan o se introduzcan.

* Caso 3: Aquí elegiría la estructura de diccionario, porque nos solicitan asignar un nombre (clave) a un número de teléfono (valor). Y esa es exactamente la estructura de un diccionario, un par clave-valor.

* Caso 4: Aquí utilizaría la estructura de conjunto, porque si nos piden comparar a los estudiantes que pertenezcan a dos grupos, la operación intersección "&" nos arrojaría los estudiantes directamente.

* Caso 5: En este caso la estructura tupla sería la indicada, porque al darnos a nosotros la imagen (por ejemplo en un archivo .jpg,.jpeg,.png,etc...) sus dimensiones son fijas, y la propiedad de inmutabilidad de la tupla es ideal para ello.

* Caso 6: Para esto utilizaría la estructura de diccionario para relacionar estos datos directamente. Adicionalmente se puede ver este par producto(clave)-precio(valor) como un registro.

#### $\hspace{6cm}\color{#c43399}{{\textbf{Ejercicio 8 — Reto integrador}}}$

```python
#Entrada y definición
grupo_A = {"Ana", "Luis", "Carlos", "María"}
grupo_B = {"Luis", "María", "Sofía", "Juan"}
notas = {
    "Ana": 4.5,
    "Luis": 3.2,
    "Carlos": 2.8,
    "María": 4.7,
    "Sofía": 3.9,
    "Juan": 2.5
}
verif_aprob_8=""
nombre_mayor=""
verif_aprob_conjun_8={}
nota_sum=0.0
prom_notas=0.0
nota_mayor=0.0

#Procesamiento y salida
print("Los estudiantes que están en ambos grupos son:", grupo_A&grupo_B)
print("Los estudiantes que están únicamente en el grupo A son:",grupo_A-grupo_B)
print("Todos los estudiantes registrados son:", grupo_A|grupo_B)
print("------------------------------------")
for i in notas.values(): #Reciclaje del código anterior "ejercicio #6", ya que cumple la misma función
    if (i>=3.0):
        verif_aprob_8="Aprobó"
        nota_sum=nota_sum + i 
    else:
        verif_aprob_8="No aprobó"
    verif_aprob_conjun_8[i]=verif_aprob_8

for clave_8,i in notas.items():
    if (i>nota_mayor):
        nota_mayor=i
        nombre_mayor=clave_8
    else:
        continue



prom_notas=(nota_sum)/(len(notas))

for clave,i in notas.items():
    print(clave,":",i,"-",verif_aprob_conjun_8[i])
print("El promedio de las notas registradas es:",round(prom_notas,2))
print("La nota más alta obtenida por el/la estudiante [",nombre_mayor,"] fue de:",nota_mayor)
```
