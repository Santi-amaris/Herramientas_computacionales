```python
import random


#Primera parte (definición de variables y funciones útiles antes de comenzar a ejecutar el programa)


nombre_jugador="" #Nombre del jugador que se utilizará para hacer el reporte de ronda (reporte_ronda.txt)

apuesta_inicial=0 #Apuesta inicial que introduce el jugador

ganancia_jugador=0 #Ganancia del jugador al final de la ronda

cartas_baraja=["As","2 de picas","3 de picas","4 de picas","5 de picas","6 de picas","7 de picas","8 de picas","9 de picas","10 de picas","Joker de picas","Reina de picas","Rey de picas"] #Lista de las cartas disponibles a ser elegidas, elegibles por índice

cartas_baraja_puntos={
    "As":1,
    "2 de picas":2,
    "3 de picas":3,
    "4 de picas":4,
    "5 de picas":5,
    "6 de picas":6,
    "7 de picas":7,
    "8 de picas":8,
    "9 de picas":9,
    "10 de picas":10,
    "Joker de picas":10,
    "Reina de picas":10,
    "Rey de picas":10
}

carta_sacada_jugador="" #Variable que guarda la carta que saca turno a turno el jugador

carta_sacada_puntos_jugador=0 #Valor de la carta sacada turno a turno del jugador

carta_sacada_dealer="" #Variable que guarda la carta que saca turno a turno el dealer

carta_sacada_puntos_dealer=0 #Valor de la carta sacada turno a turno del dealer

puntuacion_jugador=0 #Puntuación del jugador turno a turno

puntuacion_dealer=0 #Puntuación del "Dealer" turno a turno

contador_turno=1 #Contador de los turnos

contador_doblar_apuesta=0

opcion_accion=0 #Opcion que le permite al jugador elegir entre las (4) opciones disponibles: (1) Hit, (2) Stand, (3) Double Down, (4) Salir del programa. 

condicion_salir_programa=False

condicion_ronda=True #Condición que valida que se siga jugando la ronda


#Función que define la acción (1) "Hit"/Pedir carta
def pedir_carta(puntuacion_jugador):
    
    #Carta que saca el jugador con su respectiva puntuación asociada
    
    carta_sacada_jugador=random.choice(cartas_baraja)
    
    carta_sacada_puntos_jugador=cartas_baraja_puntos[carta_sacada_jugador]
    
    puntuacion_jugador+=carta_sacada_puntos_jugador
    
    print(f'\n====Turno {contador_turno}====\n')
    print(f'==== Carta para {nombre_jugador}: {carta_sacada_jugador} ====\n')
    print(f'==== Puntuación de {nombre_jugador}: {puntuacion_jugador} ====\n')
    
    return puntuacion_jugador

#Función que define la acción (2) "Stand"/Quedarse
def quedarse():
    
    if (contador_doblar_apuesta>0):
        print(f'\n====Turno {contador_turno}====\n')
    else:
        print(f'\n====Turno {contador_turno}====\n')
        print(f'==== El jugador {nombre_jugador} decidió quedarse por el resto de la ronda ====\n')
        print(f'==== Puntuación de {nombre_jugador}: {puntuacion_jugador} ====\n')

#Función que define la acción (3) "Double Down"/Doblar la apuesta
def doblar_apuesta(contador_doblar_apuesta,apuesta_inicial,puntuacion_jugador):
    
    apuesta_inicial+=apuesta_inicial
    
    contador_doblar_apuesta+=1
    
    #Única carta que saca el jugador con su respectiva puntuación asociada después de doblar la apuesta inicial
    
    carta_sacada_jugador=random.choice(cartas_baraja)
    
    carta_sacada_puntos_jugador=cartas_baraja_puntos[carta_sacada_jugador]
    
    puntuacion_jugador+=carta_sacada_puntos_jugador
    
    print(f'\n====Turno {contador_turno}====\n')
    print(f'==== Carta para {nombre_jugador}: {carta_sacada_jugador} ====\n')
    print(f'==== Puntuación de {nombre_jugador}: {puntuacion_jugador} ====\n')
    
    return puntuacion_jugador,apuesta_inicial,contador_doblar_apuesta

#Función que define la acción (4) "Surrender"/Rendirse
def rendirse(condicion_salir_programa,ganancia_jugador):
    condicion_salir_programa=True
    ganancia_jugador-=(apuesta_inicial/2)
    print(f'\n=== El jugador {nombre_jugador} ha decidido retirarse de la mesa ===\n')
    print(f'=== Total ronda: {ganancia_jugador} ===\n')
    return condicion_salir_programa,ganancia_jugador

#Función que define el turno del dealer
def turno_dealer(contador_turno,puntuacion_dealer):
    
    #Carta que saca el dealer con su respectiva puntuación asociada
    
    carta_sacada_dealer=random.choice(cartas_baraja)
    
    carta_sacada_puntos_dealer=cartas_baraja_puntos[carta_sacada_dealer]
    
    puntuacion_dealer+=carta_sacada_puntos_dealer
    
    print(f'==== Carta para el "Dealer": {carta_sacada_dealer} ====\n')
    print(f'==== Puntuación del "Dealer": {puntuacion_dealer} ====\n')
    print(f'==== Fin del turno ====\n')
    
    contador_turno+=1
    
    return contador_turno,puntuacion_dealer


#Clase que define la cuenta bancaria del jugador

#...código futuro de la clase 







#Segunda parte (Entrada de datos, procesamiento de datos)

nombre_jugador=input(f'Bienvenido al Casino "Zarabastra" y como siempre yo seré su "Dealer de confianza". Pero antes de hacer nada...¿Me permite saber su nombre?',)

while (True):
    try:
        apuesta_inicial=float(input(f"¿Cuánto desearía apostar en esta ocasión, {nombre_jugador}? ",))
        if apuesta_inicial<=0:
            raise TypeError(f"Usted tiene que apostar una cantidad de dinero positiva {nombre_jugador}. No se admiten cantidades nulas o negativas; por favor vuelva a intentarlo...")
    except ValueError as error:
        print("Se presentó el siguiente error en consola ---",error)
        print("\nPor favor vuelva a ingresar una cantidad de dinero válida en formato númerico...\n")
    except TypeError as error:
        print("Se presentó el error ---",error)
        print("\nVuelva a ingresar una cantidad de apuesta positiva...\n")
    else:
        break


while(condicion_ronda==True):
    
    while(True):
        try:
            opcion_accion=int(input(f'¿Que acción desea tomar a continuación, {nombre_jugador}?, tenga en cuenta que su puntuación actual es de: {puntuacion_jugador} puntos. Tiene por opciones: (1)"Hit"/Pedir carta; \n(2)"Stand"/Quedarse; \n(3)"Double Down"/Doblar apuesta; \n(4)"Surrender"/Rendirse',))
            if (opcion_accion<1 or opcion_accion>4):
                raise IndexError("Usted se ha salido del intervalo válido de opciones. Por favor vuelva a ingresar una cantidad entre alguna de las opciones válidas...\n")
        except ValueError as error:
            print("Se presentó el siguiente error en consola ---",error)
            print("\nPor favor vuelva a ingresar una cantidad válida numérica entera...\n")
        except IndexError as error:
            print("Se presentó el siguiente error en consola ---",error)
            print("\nVuelva a ingresar una cantidad en el intervalo dado [1,2,3,4]...\n")
        else:
            break
    
    
    if (opcion_accion==1):
        puntuacion_jugador= pedir_carta(puntuacion_jugador) # Se ejecuta la función respectiva de la acción tomada y se actualiza y guarda la puntuación del jugador
        if (puntuacion_jugador>21):
            break
        contador_turno, puntuacion_dealer= turno_dealer(contador_turno,puntuacion_dealer) # Se ejecuta la función respectiva de la acción tomada y se actualiza y guarda la puntuación del "dealer"
        if ((puntuacion_dealer<=21 and puntuacion_dealer>=17) or puntuacion_dealer>21):
            break
        continue #línea que me asegura que terminó el turno y se realiza la siguiente iteración
        
    elif (opcion_accion==2):
        
        while(True):
            quedarse()
            contador_turno, puntuacion_dealer= turno_dealer(contador_turno,puntuacion_dealer) # Se ejecuta la función respectiva de la acción tomada y se actualiza y guarda la puntuación del "dealer"
            if ((puntuacion_dealer<=21 and puntuacion_dealer>=17) or puntuacion_dealer>21):
                condicion_ronda=False
                break
        
        continue #línea que me asegura que terminó el turno y se realiza la siguiente iteración
        
    elif (opcion_accion==3):
        puntuacion_jugador, apuesta_inicial, contador_doblar_apuesta= doblar_apuesta(contador_doblar_apuesta,apuesta_inicial,puntuacion_jugador)
        if (puntuacion_jugador>21):
            break
        while (True): #Ciclo que me asegura que al jugador ya no se le reparten más cartas y solo se queda hasta que acabe la ronda
            quedarse()
            contador_turno, puntuacion_dealer= turno_dealer(contador_turno,puntuacion_dealer)
            if ((puntuacion_dealer<=21 and puntuacion_dealer>=17) or puntuacion_dealer>21):
                condicion_ronda=False
                break
        
        continue #línea que me asegura que terminó el turno y se realiza la siguiente iteración
        
    elif (opcion_accion==4):
        
        condicion_salir_programa, ganancia_jugador= rendirse(condicion_salir_programa, ganancia_jugador)
        break
    


if (condicion_salir_programa==True):
    print(f'\n=== Esperamos volver a verlo pronto {nombre_jugador}, hasta luego... ===\n')
    print(f'=== Fin de la ronda ===\n')
elif (puntuacion_jugador>21):
    ganancia_jugador-=apuesta_inicial
    print(f'\n=== {nombre_jugador} ha perdido la ronda ===\n')
    print(f'=== Total: {ganancia_jugador} ===\n')
    print(f'=== Fin de la ronda ===\n')
elif (puntuacion_dealer>21):
    ganancia_jugador+=apuesta_inicial
    print(f'\n=== {nombre_jugador} ha ganado la ronda ===\n')
    print(f'=== Total: {ganancia_jugador} ===\n')
    print(f'=== Fin de la ronda ===\n')
elif (puntuacion_jugador>puntuacion_dealer):
    ganancia_jugador+=apuesta_inicial
    print(f'\n=== {nombre_jugador} ha ganado la ronda ===\n')
    print(f'=== Total: {ganancia_jugador} ===\n')
    print(f'=== Fin de la ronda ===\n')
elif (puntuacion_dealer>puntuacion_jugador):
    ganancia_jugador-=apuesta_inicial
    print(f'\n=== {nombre_jugador} ha perdido la ronda ===\n')
    print(f'=== Total: {ganancia_jugador} ===\n')
    print(f'=== Fin de la ronda ===\n')
elif (puntuacion_dealer==puntuacion_jugador):
    print(f'\n=== La ronda terminó en empate ===\n')
    print(f'=== Total: 0 ===\n')
    print(f'=== Fin de la ronda ===\n')
```
