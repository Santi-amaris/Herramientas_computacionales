## 1. ¿Por qué trabajar con archivos?

Hasta ahora hemos trabajado principalmente con datos almacenados en variables, listas, diccionarios, etc.

El problema es que estos datos normalmente desaparecen cuando termina la ejecución del programa.

Un **archivo** permite almacenar información de forma persistente para poder utilizarla posteriormente.

### Ejemplo

Supongamos que tenemos:

```python
notas = [4.5, 3.8, 4.2]
```

Mientras el programa está ejecutándose, podemos utilizar `notas`. Pero si cerramos Python, la variable deja de existir.

Podemos guardar la información en un archivo:

```text
4.5
3.8
4.2
```

y posteriormente leerla desde Python.

### Idea fundamental

**Programa → procesa datos → archivo**

y posteriormente:

**archivo → programa → procesa datos**

## 2. Archivos de texto

Un archivo de texto contiene caracteres organizados en líneas.

Algunos ejemplos:

- `.txt`
- `.csv`
- `.log`
- archivos de configuración sencillos

Por ahora trabajaremos principalmente con archivos `.txt`.

### Estructura conceptual

```text
proyecto/
│
├── archivos/

│   └── resultados.txt

```

## 3. Abrir un archivo: `open()`

La función básica para trabajar con archivos es:

```python
open(nombre_archivo, modo)
```

Por ejemplo:

```python
archivo = open("datos.txt", "r")
```

El segundo argumento indica el **modo** de apertura.

### Modos principales

| Modo | Significado |
|---|---|
| `"r"` | Leer |
| `"w"` | Escribir. Crea el archivo o reemplaza su contenido |
| `"a"` | Agregar contenido al final |
| `"x"` | Crear un archivo nuevo; falla si ya existe |

También podemos especificar la codificación:

```python
open("datos.txt", "r", encoding="utf-8")
```

Para archivos de texto, utilizar explícitamente `encoding="utf-8"` es una buena práctica.

## 4. Primera escritura de un archivo

Vamos a crear un archivo de texto.

```python
archivo = open("saludo.txt", "w", encoding="utf-8")
archivo.write("Hola, mundo")
archivo.close()
```

### ¿Qué ocurrió?

1. `open()` abrió o creó el archivo.
2. `"w"` indica que queremos escribir.
3. `write()` escribió una cadena.
4. `close()` cerró el archivo.

### ⚠️ Importante

El modo `"w"` puede **sobrescribir el contenido existente**.

Por ejemplo:

```python
archivo = open("saludo.txt", "w", encoding="utf-8")
archivo.write("Nuevo contenido")
archivo.close()
```

Si `saludo.txt` ya tenía información, esta será reemplazada.

````python
archivo = open("saludo.txt", "w", encoding="utf-8")
archivo.write("Hola, mundo")
archivo.close()

print("Archivo creado.")
````

## 5. Escribir varias líneas

`write()` recibe una cadena. Si queremos escribir varias líneas, debemos incluir `\n`.

```python
archivo.write("Ana\n")
archivo.write("Carlos\n")
archivo.write("Laura\n")
```

También podemos utilizar una sola cadena:

```python
texto = "Ana\nCarlos\nLaura\n"
archivo.write(texto)
```

Resultado:


No se agrega automáticamente un salto de línea.

````python
archivo = open("estudiantes.txt", "w", encoding="utf-8")

archivo.write("Ana")
archivo.write("Carlos")
archivo.write("Laura")

archivo.close()

print("Datos guardados.")
````

## 6. El método `close()`

Cuando usamos directamente `open()`, debemos cerrar el archivo:

```python
archivo = open("datos.txt", "r")
# operaciones
archivo.close()
```

Cerrar el archivo es importante porque libera recursos y asegura que las operaciones pendientes se completen.

Sin embargo, existe una forma más segura y conveniente de trabajar con archivos:

# `with`

Esto nos lleva al concepto de **context manager**.

## 7. Context managers: `with`

La estructura recomendada para trabajar con archivos es:

```python
with open("datos.txt", "r", encoding="utf-8") as archivo:
    # trabajar con archivo
```

Python se encarga de cerrar el archivo al salir del bloque `with`.

### Comparación

#### Forma manual

```python
archivo = open("datos.txt", "r", encoding="utf-8")

contenido = archivo.read()

archivo.close()
```

#### Con `with`

```python
with open("datos.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
```

La segunda forma es normalmente preferible.

### Idea conceptual

```text
ENTRAR AL CONTEXTO
       ↓
abrir archivo
       ↓
realizar operaciones
       ↓
SALIR DEL CONTEXTO
       ↓
cerrar archivo automáticamente
```

````python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
````

## 8. Lectura completa: `read()`

El método:

```python
archivo.read()
```

lee todo el contenido disponible y lo devuelve como una cadena.

Ejemplo:

```python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
```

### ¿Qué tipo de dato obtenemos?

```python
type(contenido)
```

El resultado es:

```text
str
```

Por tanto, después podemos utilizar los métodos de cadenas que ya conocemos.

````python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(type(contenido))
print(contenido)
````

## 9. Lectura por líneas: `readline()`

`readline()` lee una línea a la vez.

```python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    linea = archivo.readline()
    print(linea)
```

Podemos repetir la operación:

```python
linea1 = archivo.readline()
linea2 = archivo.readline()
```

### Observación

Cada llamada avanza la posición dentro del archivo.

````python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    linea1 = archivo.readline()
    linea2 = archivo.readline()

print("Primera línea:", linea1)
print("Segunda línea:", linea2)
````

## 10. Lectura de todas las líneas: `readlines()`

El método:

```python
archivo.readlines()
```

devuelve una lista donde cada elemento corresponde a una línea.

```python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()

print(lineas)
```

Podemos obtener algo como:

```python
['Ana\n', 'Carlos\n', 'Laura\n']
```

### Atención

El `\n` forma parte de las cadenas leídas.

Podemos eliminarlo utilizando:

```python
linea.strip()
```

````python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()

print(lineas)

print("Sin saltos de línea:")
for linea in lineas:
    print(linea.strip())
````

## 11. La forma más natural: recorrer el archivo

Una de las formas más útiles de procesar un archivo de texto es recorrerlo directamente:

```python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea.strip())
```

Esto evita tener que cargar necesariamente todas las líneas en una lista.

### Patrón importante

```python
with open("archivo.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        # procesar linea
```

Este patrón será muy utilizado en programas reales.

````python
with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        nombre = linea.strip()
        print("Estudiante:", nombre)
````

## 12. Agregar información: modo `"a"`

El modo `"a"` significa **append**: agregar al final del archivo.

```python
with open("estudiantes.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Pedro\n")
```

El contenido anterior se conserva.

### Diferencia fundamental

```text
"w" → reemplaza el contenido
"a" → conserva y agrega al final
```

Esta diferencia es muy importante para evitar perder información accidentalmente.

````python
with open("estudiantes.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Pedro")

with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````

# 14. Organización de datos en archivos

Un archivo puede almacenar datos de diferentes formas.

Por ejemplo, tenemos estudiantes con nombre y edad.

Una opción poco estructurada sería:

```text
Ana
20
Carlos
21
Laura
19
```

Es posible interpretarlo, pero el programa debe asumir que cada dos líneas forman un registro.

Una alternativa más organizada es utilizar un separador:

```text
Ana,20
Carlos,21
Laura,19
```

Ahora cada línea representa un registro:

```text
nombre,edad
```

Este concepto será fundamental cuando trabajemos posteriormente con archivos CSV.

````python
with open("personas.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Ana,20\n")
    archivo.write("Carlos,21\n")
    archivo.write("Laura,19\n")

print("Archivo creado correctamente.")
````

## 15. Separar información con `split()`

Si tenemos:

```python
linea = "Ana,20"
```

podemos utilizar:

```python
datos = linea.split(",")
```

Resultado:

```python
['Ana', '20']
```

Ahora podemos separar los campos:

```python
nombre = datos[0]
edad = int(datos[1])
```

### Patrón general

```python
with open("personas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        linea = linea.strip()
        datos = linea.split(",")

        nombre = datos[0]
        edad = int(datos[1])

        print(nombre, edad)
```

````python
with open("personas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        linea = linea.strip()

        # Separar los campos utilizando la coma
        datos = linea.split(",")

        # El primer campo corresponde al nombre
        nombre = datos[0]

        # El segundo campo corresponde a la edad
        edad = int(datos[1])

        print(f"{nombre} tiene {edad} años.")
````

# 16. Actividad guiada 2 — Procesar notas

Vamos a utilizar un archivo con la siguiente estructura:

```text
Ana,4.5
Carlos,3.8
Laura,4.2
Pedro,2.9
```

El objetivo será:

1. Leer el archivo.
2. Separar nombre y nota.
3. Convertir la nota a `float`.
4. Mostrar cada estudiante.
5. Calcular el promedio.
6. Contar cuántos estudiantes aprobaron.

Supongamos que la nota mínima para aprobar es `3.0`.

### Antes de programar

Preguntar a los estudiantes:

- ¿Qué tipo de dato es la línea completa?
- ¿Qué devuelve `split(",")`?
- ¿Por qué necesitamos convertir la nota?
- ¿Cómo podemos acumular el promedio?
- ¿Cómo podemos contar aprobados?

````python
with open("notas.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Ana,4.5\n")
    archivo.write("Carlos,3.8\n")
    archivo.write("Laura,4.2\n")
    archivo.write("Pedro,2.9\n")

print("Archivo de notas creado.")
````

````python
suma = 0
cantidad = 0
aprobados = 0

with open("notas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        linea = linea.strip()

        if linea:
            datos = linea.split(",")

            nombre = datos[0]
            nota = float(datos[1])

            print(f"{nombre}: {nota}")

            suma += nota
            cantidad += 1

            if nota >= 3.0:
                aprobados += 1

promedio = suma / cantidad

print("Resultados")
print("Promedio:", promedio)
print("Aprobados:", aprobados)
````

## 17. El uso de `strip()`

Una operación muy frecuente será:

```python
linea = linea.strip()
```

`strip()` elimina espacios y saltos de línea al inicio y al final.

Por ejemplo:

```python
linea = "   Ana,4.5\n"
```

Después de:

```python
linea.strip()
```

obtenemos:

```text
Ana,4.5
```

### ¿Por qué es útil?

Porque al leer un archivo normalmente encontramos:

```python
'Ana,4.5\n'
```

y queremos procesar:

```python
'Ana,4.5'
```

### Cuidado

`strip()` no debe confundirse con `split()`:

- `strip()` limpia extremos.
- `split()` divide una cadena en partes.

# 18. Escribir datos generados por un programa

También podemos guardar resultados calculados.

Por ejemplo:

```python
with open("resultado.txt", "w", encoding="utf-8") as archivo:
    archivo.write(f"Promedio: {promedio}\n")
    archivo.write(f"Aprobados: {aprobados}\n")
```

Así podemos construir un flujo completo:

```text
archivo de entrada
       ↓
lectura
       ↓
procesamiento
       ↓
cálculos
       ↓
archivo de resultados
```

Este patrón aparece constantemente en programación científica, análisis de datos y automatización.

````python
suma = 0
cantidad = 0
aprobados = 0

with open("notas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        linea = linea.strip()

        if linea:
            nombre, nota = linea.split(",")
            nota = float(nota)

            suma += nota
            cantidad += 1

            if nota >= 3.0:
                aprobados += 1

promedio = suma / cantidad

with open("resultado.txt", "w", encoding="utf-8") as archivo:
    archivo.write("REPORTE DE NOTAS")
    archivo.write("================")
    archivo.write(f"Promedio: {promedio:.2f}")
    archivo.write(f"Aprobados: {aprobados}")

print("Reporte generado.")
````

# 19. Errores frecuentes

### 1. El archivo no existe

```python
with open("datos.txt", "r") as archivo:
```

puede producir:

```text
FileNotFoundError
```

Si estamos leyendo, el archivo debe existir.

### 2. Ruta incorrecta

Python interpreta las rutas relativas respecto al directorio de trabajo actual.

Podemos revisar dónde estamos trabajando con:

```python
import os
print(os.getcwd())
```

### 3. Sobrescribir accidentalmente un archivo

```python
open("datos.txt", "w")
```

puede borrar el contenido anterior.

### 4. Olvidar convertir tipos

```python
edad = datos[1]
```

produce una cadena.

Si necesitamos un entero:

```python
edad = int(datos[1])
```

### 5. No limpiar el salto de línea

Si usamos directamente:

```python
linea.split(",")
```

podemos terminar con un último elemento que contiene `\n`.

Por eso suele ser conveniente:

```python
linea = linea.strip()
```

# 21. Actividad integradora en clase

## Registro de temperaturas

Construiremos un programa que trabaje con:

```text
temperaturas.txt
```

Cada línea tendrá:

```text
ciudad,temperatura
```

Por ejemplo:

```text
Bogota,16.5
Medellin,24.3
Cali,25.1
Cartagena,29.8
```

### El programa debe

1. Crear el archivo con los datos iniciales.
2. Leer los registros.
3. Mostrar cada ciudad y temperatura.
4. Calcular la temperatura promedio.
5. Encontrar la temperatura máxima.
6. Encontrar la temperatura mínima.
7. Crear un archivo `reporte_temperaturas.txt`.
8. Guardar allí los resultados.

### Pistas

Para el máximo y mínimo podemos utilizar:

```python
max()
min()
```

Pero primero debemos construir una lista de temperaturas.

Para el promedio:

```python
sum(temperaturas) / len(temperaturas)
```

### Extensión

Agregar una nueva ciudad al archivo utilizando el modo `"a"` y volver a calcular el reporte.

````python
# Solución guiada / referencia

with open("temperaturas.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Bogota,16.5\n")
    archivo.write("Medellin,24.3\n")
    archivo.write("Cali,25.1\n")
    archivo.write("Cartagena,29.8\n")


ciudades = []
temperaturas = []

with open("temperaturas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        linea = linea.strip()

        if linea:
            ciudad, temperatura = linea.split(",")
            temperatura = float(temperatura)

            ciudades.append(ciudad)
            temperaturas.append(temperatura)

promedio = sum(temperaturas) / len(temperaturas)
maxima = max(temperaturas)
minima = min(temperaturas)

print("Temperaturas:")
for ciudad, temperatura in zip(ciudades, temperaturas):
    print(f"{ciudad}: {temperatura} °C")

print(f"Promedio: {promedio:.2f} °C")
print(f"Máxima: {maxima:.2f} °C")
print(f"Mínima: {minima:.2f} °C")


with open("reporte_temperaturas.txt", "w", encoding="utf-8") as archivo:
    archivo.write("REPORTE DE TEMPERATURAS")
    archivo.write("=======================")
    archivo.write(f"Promedio: {promedio:.2f} °C")
    archivo.write(f"Máxima: {maxima:.2f} °C")
    archivo.write(f"Mínima: {minima:.2f} °C")

print("Reporte guardado en reporte_temperaturas.txt")
````
