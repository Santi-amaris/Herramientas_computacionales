# 1. ¿Qué es un error?

Un programa puede fallar por diferentes razones. No todos los errores son del mismo tipo.

En términos generales podemos distinguir:

- **Errores de sintaxis:** Python no puede interpretar el código.
- **Errores de ejecución:** el programa comienza a ejecutarse, pero ocurre un problema durante la ejecución.
- **Errores lógicos:** el programa se ejecuta, pero produce un resultado incorrecto.

Una idea importante:

> Un programa que no genera errores de sintaxis o de ejecución no necesariamente es un programa correcto.

Los errores lógicos pueden ser especialmente difíciles de detectar porque Python no necesariamente muestra un mensaje de error.


## 1.1 Errores de sintaxis

Ocurren cuando el código no cumple las reglas de escritura de Python.

Ejemplo:

```python
if edad >= 18
    print("Mayor de edad")
```

Falta `:` después de la condición.

Python detendrá la ejecución antes de comenzar el programa.

## 1.3 Errores lógicos

Son errores en el razonamiento o algoritmo.

El programa se ejecuta, pero el resultado no es el esperado.

Ejemplo:


No existe un error de sintaxis ni una excepción, pero la fórmula utilizada para el área de un rectángulo es incorrecta.

# 2. Excepciones en Python

Una **excepción** es un evento que ocurre durante la ejecución de un programa y que interrumpe el flujo normal de instrucciones.

Por ejemplo:

```python
numero = int("hola")
```

Python intenta convertir `"hola"` a un entero, pero no puede hacerlo.

Se genera:

```text
ValueError
```

La pregunta es:

> ¿Podemos evitar que nuestro programa simplemente se detenga?

Sí. Podemos **manejar la excepción** utilizando `try` y `except`.


# 3. `try` y `except`

La estructura básica es:

```python
try:
    # Código que podría generar una excepción
except:
    # Código que se ejecuta si ocurre una excepción
```

El bloque `try` contiene el código que queremos intentar ejecutar.

El bloque `except` indica qué hacer si ocurre una excepción.

### Ejemplo


````python
try:
    numero = int(input("Ingrese un número entero: "))
    print("Número ingresado:", numero)
except:
    print("La entrada no es válida.")
````

### Pregunta para los estudiantes

Antes de ejecutar:

1. ¿Qué ocurre si el usuario escribe `25`?
2. ¿Qué ocurre si escribe `hola`?
3. ¿Por qué el programa no se detiene en el segundo caso?

### Idea clave

`try` **no significa que el código sea correcto**.

Significa:

> "Intenta ejecutar este código y, si ocurre una excepción que pueda manejarse, ejecuta el bloque correspondiente."


# 4. Capturar excepciones específicas

Aunque podemos escribir:

```python
except:
```

no suele ser la mejor práctica.

Es preferible especificar qué excepción esperamos manejar.

```python
try:
    numero = int(input("Número: "))
except ValueError:
    print("Debe ingresar un número entero.")
```

Esto permite diferenciar problemas distintos.

### Ejemplo


````python
try:
    a = int(input("Ingrese el numerador: "))
    b = int(input("Ingrese el denominador: "))
    resultado = a / b
    print("Resultado:", resultado)

except ValueError:
    print("Debe ingresar números enteros.")

except ZeroDivisionError:
    print("El denominador no puede ser cero.")
````

### ¿Por qué usar excepciones específicas?

Supongamos que el usuario escribe:

```text
10
0
```

La conversión a entero funciona, pero la división genera `ZeroDivisionError`.

Si escribe:

```text
diez
2
```

la conversión genera `ValueError`.

Podemos ofrecer un mensaje diferente para cada situación.

Esto hace que nuestros programas sean:

- más claros,
- más robustos,
- más fáciles de mantener,
- más útiles para el usuario.


# 5. La excepción como objeto

Podemos guardar la información de la excepción utilizando `as`.

```python
try:
    numero = int("hola")
except ValueError as error:
    print(error)
```

La variable `error` contiene información sobre la excepción.

También podemos consultar su tipo:

```python
print(type(error))
```


````python
try:
    numero = int("hola")
except ValueError as error:
    print("Se produjo una excepción.")
    print("Mensaje:", error)
    print("Tipo:", type(error))
````

# 6. `else`

Python también permite utilizar `else`.

La estructura es:

```python
try:
    # Código que puede generar una excepción
except TipoDeError:
    # Si ocurre la excepción
else:
    # Si NO ocurre ninguna excepción
```

El bloque `else` se ejecuta solamente cuando el bloque `try` termina correctamente.

### Ejemplo


````python
try:
    numero = int(input("Ingrese un número entero: "))
except ValueError:
    print("Entrada inválida.")
else:
    print("Conversión exitosa.")
    print("El número es:", numero)
````

## ¿Por qué utilizar `else`?

Permite separar claramente:

- el código que podría generar una excepción,
- el manejo de la excepción,
- el código que debe ejecutarse únicamente si todo salió bien.

Por ejemplo:

```python
try:
    edad = int(input("Edad: "))
except ValueError:
    print("Edad inválida.")
else:
    if edad >= 18:
        print("Es mayor de edad.")
```


# 7. `finally`

El bloque `finally` se ejecuta **siempre**, independientemente de si ocurrió una excepción o no.

Estructura:

```python
try:
    ...
except:
    ...
finally:
    ...
```

Es especialmente útil para tareas de limpieza o acciones que deben realizarse al terminar.

### Ejemplo sencillo


````python
try:
    numero = int(input("Ingrese un número: "))
    resultado = 100 / numero
    print("Resultado:", resultado)

except ValueError:
    print("Debe ingresar un número.")

except ZeroDivisionError:
    print("No se puede dividir entre cero.")

finally:
    print("Fin del proceso.")
````

### Estructura completa

Podemos combinar `try`, `except`, `else` y `finally`:

```python
try:
    # Intentar realizar una operación

except TipoDeError:
    # Manejar una excepción

else:
    # Ejecutar si no hubo excepciones

finally:
    # Ejecutar siempre
```

### Orden de ejecución

Normalmente:

```text
try
 ↓
¿Ocurrió una excepción?
 ├── Sí → except → finally
 └── No → else → finally
```


# 8. `raise`: generar excepciones intencionalmente

Hasta ahora hemos manejado excepciones que Python genera automáticamente.

También podemos generar una excepción nosotros mismos usando `raise`.

Sintaxis:

```python
raise TipoDeExcepcion("Mensaje")
```

Ejemplo:

```python
edad = -5

if edad < 0:
    raise ValueError("La edad no puede ser negativa.")
```

Esto es útil para **validar datos** y establecer reglas en nuestros programas.


````python
edad = -5

if edad < 0:
    raise ValueError("La edad no puede ser negativa.")

print("Edad válida:", edad)
````

## 8.1 `raise` dentro de una función

Una función puede validar sus argumentos y generar una excepción si no cumplen las condiciones esperadas.

Ejemplo:

```python
def calcular_promedio(notas):
    if len(notas) == 0:
        raise ValueError("La lista de notas no puede estar vacía.")

    return sum(notas) / len(notas)
```

La función establece explícitamente qué condiciones considera inválidas.


````python
def calcular_promedio(notas):
    if len(notas) == 0:
        raise ValueError("La lista de notas no puede estar vacía.")

    return sum(notas) / len(notas)


print(calcular_promedio([4.0, 4.5, 3.8]))
````

### Combinar `raise` con `try` / `except`

Podemos generar una excepción en una función y manejarla desde el programa principal:

```python
try:
    resultado = calcular_promedio([])
except ValueError as error:
    print("No fue posible calcular el promedio.")
    print(error)
```

Esta separación es importante:

- La función detecta el problema.
- El código que utiliza la función decide cómo responder.


````python
def calcular_promedio(notas):
    if len(notas) == 0:
        raise ValueError("La lista de notas no puede estar vacía.")

    return sum(notas) / len(notas)


try:
    promedio = calcular_promedio([])
    print("Promedio:", promedio)
except ValueError as error:
    print("Error:", error)
````

# 10. Buenas prácticas para manejar excepciones

### 1. Capturar excepciones específicas

Preferir:

```python
except ValueError:
```

en lugar de:

```python
except:
```

### 2. No ocultar los errores

Evitar:

```python
try:
    ...
except:
    pass
```

Esto puede hacer que un problema pase desapercibido.

### 3. Mostrar mensajes útiles

Un buen mensaje debería ayudar a entender qué ocurrió y, cuando sea posible, cómo solucionarlo.

### 4. Mantener el `try` pequeño

Es preferible colocar dentro del `try` solamente las instrucciones que realmente pueden generar la excepción que queremos manejar.

### 5. Usar `raise` para validar reglas

Si una función recibe datos inválidos, puede generar una excepción explícitamente.


# 11. Ejemplo integrador: calculadora segura

Vamos a construir una pequeña calculadora que:

1. Solicite dos números.
2. Solicite una operación.
3. Maneje entradas inválidas.
4. Evite la división entre cero.
5. Informe al usuario qué ocurrió.


````python
def calcular(a, b, operacion):
    if operacion == "+":
        return a + b
    elif operacion == "-":
        return a - b
    elif operacion == "*":
        return a * b
    elif operacion == "/":
        if b == 0:
            raise ZeroDivisionError("No se puede dividir entre cero.")
        return a / b
    else:
        raise ValueError("Operación no válida.")


try:
    a = float(input("Primer número: "))
    b = float(input("Segundo número: "))
    operacion = input("Operación (+, -, *, /): ")

    resultado = calcular(a, b, operacion)

except ValueError as error:
    print("Error de valor:", error)

except ZeroDivisionError as error:
    print("Error matemático:", error)

else:
    print("Resultado:", resultado)

finally:
    print("Fin de la calculadora.")
````

## 12.1 Puntos de interrupción

Un breakpoint indica a Python:

> "Cuando llegues a esta línea, detén temporalmente la ejecución."

En VS Code:

1. Abra un archivo `.py`.
2. Haga clic a la izquierda del número de línea.
3. Aparecerá un punto rojo.
4. Ejecute el programa con el depurador.
5. Cuando el programa llegue a esa línea, la ejecución se pausará.

### Ejemplo

Crear un archivo llamado:

```text
debug_ejemplo.py
```

con:


````python
def calcular_total(precio, cantidad):
    subtotal = precio * cantidad
    descuento = subtotal * 0.10
    total = subtotal - descuento
    return total


precio = 100
cantidad = 3

total = calcular_total(precio, cantidad)

print("Total:", total)
````

## 12.2 Ejecución paso a paso

Las herramientas principales son:

### Continue

Continúa la ejecución hasta el siguiente breakpoint.

### Step Over

Ejecuta la línea actual y pasa a la siguiente.

### Step Into

Entra dentro de una función que se está llamando.

### Step Out

Sale de la función actual y vuelve al punto desde donde fue llamada.

### Idea fundamental

La depuración permite observar **cómo cambian las variables durante la ejecución**, no solamente el resultado final.
