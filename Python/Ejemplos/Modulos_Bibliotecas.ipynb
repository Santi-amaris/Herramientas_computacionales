## 1. ¿Por qué necesitamos módulos?

Imaginemos un programa que calcula propiedades físicas. En un único archivo podríamos terminar teniendo:

- funciones para conversiones de unidades,
- funciones matemáticas,
- funciones para leer datos,
- funciones para imprimir resultados,
- cálculos principales,
- código de prueba.

El programa puede funcionar, pero rápidamente se vuelve difícil de leer y mantener.

Una alternativa es separar responsabilidades:

```text
proyecto/
├── main.py
├── conversiones.py
├── calculos.py
└── utilidades.py
```

Cada archivo puede contener código relacionado con una tarea específica.

### Ventajas

- **Organización:** cada archivo tiene una responsabilidad clara.
- **Reutilización:** una función puede utilizarse desde diferentes programas.
- **Mantenimiento:** es más sencillo encontrar y corregir errores.
- **Colaboración:** diferentes personas pueden trabajar en diferentes partes.
- **Abstracción:** podemos utilizar funciones sin necesitar conocer todos sus detalles internos.



## 2. ¿Qué es un módulo?

En Python, un **módulo** es, en términos sencillos, un archivo `.py` que contiene código Python que podemos reutilizar.

Por ejemplo, si tenemos:

```text
matematicas.py
```

y dentro escribimos:

```python
PI = 3.141592653589793

def area_circulo(radio):
    return PI * radio**2
```

entonces `matematicas.py` es un módulo.

Otro archivo puede importar ese módulo y utilizar sus elementos.

### Una distinción importante

- **Módulo:** normalmente un archivo `.py`.
- **Paquete:** una estructura que agrupa módulos.
- **Biblioteca:** conjunto de módulos y/o paquetes que proporcionan funcionalidades reutilizables.

En esta clase trabajaremos principalmente con módulos.



## 3. Importación de módulos

La forma básica de importar un módulo es:

```python
import modulo
```

Después accedemos a sus elementos utilizando:

```python
modulo.elemento
```

Por ejemplo:


````python
import math

print(math.pi)
print(math.sqrt(25))
print(math.sin(math.pi / 2))
````


### ¿Qué está ocurriendo?

Con:

```python
import math
```

le indicamos a Python que queremos utilizar el módulo `math`.

Luego:

```python
math.sqrt(25)
```

significa:

> Utilizar la función `sqrt` que pertenece al módulo `math`.

Esta notación ayuda a evitar confusiones entre funciones con el mismo nombre que puedan pertenecer a módulos diferentes.



## 4. Importar solamente algunos elementos

También podemos importar elementos específicos:

```python
from math import sqrt, pi
```

Ahora podemos utilizar directamente `sqrt` y `pi`.


````python
from math import sqrt, pi

radio = 3
area = pi * radio**2

print("Área:", area)
print("Raíz de 81:", sqrt(81))
````


### Comparación

**Forma 1:**

```python
import math
math.sqrt(25)
```

**Forma 2:**

```python
from math import sqrt
sqrt(25)
```

Las dos son válidas.

Para estudiantes que están comenzando, `import math` suele ser una forma muy clara de conservar la relación entre la función y el módulo.



## 5. Alias

Podemos asignar un alias a un módulo:

```python
import math as m
```

y luego utilizar:

```python
m.sqrt(25)
```

Los alias son especialmente comunes con algunas bibliotecas científicas. Por ejemplo:

```python
import numpy as np
import pandas as pd
```

El alias no cambia el módulo. Solamente cambia el nombre que utilizamos para referirnos a él dentro de nuestro programa.


````python
import math as m

angulo = m.pi / 4
print("cos(pi/4) =", m.cos(angulo))
````


## 6. Una biblioteca fundamental: `math`

El módulo `math` proporciona funciones y constantes matemáticas.

Algunos ejemplos:

| Elemento | Función |
|---|---|
| `math.pi` | Número π |
| `math.e` | Número e |
| `math.sqrt(x)` | Raíz cuadrada |
| `math.sin(x)` | Seno |
| `math.cos(x)` | Coseno |
| `math.tan(x)` | Tangente |
| `math.log(x)` | Logaritmo natural |
| `math.exp(x)` | Exponencial |
| `math.factorial(n)` | Factorial |

**Importante:** las funciones trigonométricas de `math` trabajan con ángulos en **radianes**.


````python
import math

angulo_grados = 30
angulo_radianes = math.radians(angulo_grados)

print("Ángulo en radianes:", angulo_radianes)
print("sin(30°) =", math.sin(angulo_radianes))
print("cos(30°) =", math.cos(angulo_radianes))
````

## 8. Módulo `random`

`random` permite generar valores pseudoaleatorios.

Ejemplos:

```python
random.randint(a, b)
```

genera un entero entre `a` y `b`, incluyendo ambos extremos.

También podemos seleccionar aleatoriamente un elemento de una lista.


````python
import random

print("Número aleatorio:", random.randint(1, 10))

colores = ["rojo", "azul", "verde", "amarillo"]
print("Color seleccionado:", random.choice(colores))
````

## 9. Módulo `statistics`

Podemos utilizar funciones estadísticas básicas sin programarlas desde cero.

Algunas funciones:

- `mean()`: media.
- `median()`: mediana.
- `mode()`: moda.
- `stdev()`: desviación estándar muestral.


````python
import statistics

datos = [10, 12, 13, 15, 15, 18, 20]

print("Media:", statistics.mean(datos))
print("Mediana:", statistics.median(datos))
print("Moda:", statistics.mode(datos))
````


## 11. Importaciones y espacios de nombres

Consideremos:

```python
import math
import random
```

Python mantiene nombres asociados a cada módulo:

```text
math.sqrt
math.pi

random.randint
random.choice
```

Esto se conoce como trabajar mediante un **espacio de nombres**.

Una razón para utilizar:

```python
math.sqrt()
```

en lugar de importar todo directamente es que deja claro de dónde proviene `sqrt`.

### Importación que debemos evitar

```python
from math import *
```

Esta forma importa prácticamente todos los nombres del módulo al espacio actual y puede generar confusión o conflictos de nombres.

Para este curso preferiremos:

```python
import math
```

o importaciones específicas:

```python
from math import sqrt, pi
```



## 12. Crear nuestro propio módulo

Ahora vamos a construir un módulo.

Supongamos que tenemos un proyecto:


## 15. `if __name__ == "__main__"`

Cuando un archivo Python puede funcionar como módulo y también como programa independiente, es útil separar el código de prueba o ejecución principal.

Ejemplo:

```python
def cuadrado(x):
    return x**2

if __name__ == "__main__":
    print(cuadrado(5))
```

Cuando ejecutamos directamente ese archivo:

```bash
python archivo.py
```

el bloque se ejecuta.

Cuando importamos el archivo desde otro programa, el bloque principal no se ejecuta automáticamente.

### ¿Por qué es útil?

Permite que un mismo archivo contenga:

- funciones reutilizables,
- y código para probarlas o ejecutar un ejemplo.

Esta estructura será especialmente útil cuando nuestros programas comiencen a crecer.

### Regla práctica

No debemos crear módulos simplemente porque podemos.

La pregunta útil es:

> ¿Tiene sentido que este conjunto de funciones y datos esté agrupado porque tiene una responsabilidad común?

## 19. Errores frecuentes

### Error 1: el nombre del módulo no coincide

Si el archivo se llama:

```text
conversiones.py
```

debemos escribir:

```python
import conversiones
```

y no:

```python
import conversion
```

### Error 2: archivo en una carpeta incorrecta

Python debe poder encontrar el módulo que queremos importar.

### Error 3: olvidar el prefijo del módulo

Si escribimos:

```python
import math
math.sqrt(16)
```

funciona.

Pero:

```python
import math
sqrt(16)
```

no funciona porque `sqrt` no fue importado directamente.

### Error 4: confundir `print()` con `return`

Un módulo debe contener funciones que **devuelvan resultados** cuando esos resultados serán utilizados por otras partes del programa.

### Error 5: poner nombres problemáticos a los archivos

Evitar nombres como:

```text
math.py
random.py
statistics.py
```

si queremos importar los módulos estándar correspondientes. Un archivo local con esos nombres puede interferir con los imports.
