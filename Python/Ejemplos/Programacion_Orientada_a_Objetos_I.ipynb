## 1. ¿Por qué necesitamos la programación orientada a objetos?

Hasta ahora hemos trabajado con variables, tipos de datos, condicionales, ciclos, funciones y colecciones. Con estas herramientas podemos resolver muchos problemas. Sin embargo, cuando un programa representa entidades con varias características y comportamientos, puede ser útil organizar el código de otra manera.

Imagina un programa para administrar libros de una biblioteca. De cada libro nos interesa guardar:

- Título
- Autor
- Año de publicación
- Disponibilidad

También podríamos necesitar acciones como prestar el libro, devolverlo o mostrar su información.

Una solución inicial podría utilizar variables independientes y funciones. La programación orientada a objetos (POO) permite **agrupar datos relacionados y las operaciones que trabajan con ellos** en una misma estructura: el objeto.

La POO es una forma de organizar programas. No reemplaza las variables ni las funciones; ofrece una manera adicional de estructurar el código cuando el problema lo requiere.

### Idea central

Un objeto suele combinar:

- **Estado:** los datos que guarda (sus atributos).
- **Comportamiento:** las acciones que puede realizar (sus métodos).


````python
# Una representación sin clases: los datos están separados
titulo = "Cien años de soledad"
autor = "Gabriel García Márquez"
disponible = True

def mostrar_libro(titulo, autor, disponible):
    estado = "Disponible" if disponible else "Prestado"
    print(f"{titulo} — {autor}. Estado: {estado}")

mostrar_libro(titulo, autor, disponible)
````

**Observación:** este código funciona. Pero si tenemos cientos de libros, tendremos que gestionar muchos grupos de datos y recordar qué variables corresponden a cada libro. Una clase nos permite definir una estructura reutilizable para crear objetos con la misma clase de información y comportamiento.

## 2. Conceptos fundamentales

### 2.1 Clase

Una **clase** es una definición o plantilla que describe qué datos y comportamientos tendrán los objetos de cierto tipo.

Por ejemplo, una clase `Libro` puede definir que cada libro tenga título, autor y disponibilidad, además de un método para mostrar sus datos.

### 2.2 Objeto

Un **objeto** es una instancia concreta de una clase. Si `Libro` es la plantilla, un libro específico —con un título y autor particulares— es un objeto.

Podemos crear muchos objetos a partir de una misma clase, cada uno con sus propios datos.

### 2.3 Atributo

Un **atributo** es un dato asociado a un objeto. En un libro, `titulo`, `autor` y `disponible` pueden ser atributos.

### 2.4 Método

Un **método** es una función definida dentro de una clase. Describe una operación relacionada con los objetos de esa clase. Por ejemplo, `mostrar_informacion()` o `prestar()`.

| Concepto | Pregunta que responde | Ejemplo |
|---|---|---|
| Clase | ¿Qué tipo de entidad estamos definiendo? | `Libro` |
| Objeto | ¿Cuál entidad concreta tenemos? | El ejemplar de *Cien años de soledad* |
| Atributo | ¿Qué información guarda? | `titulo`, `autor` |
| Método | ¿Qué acción puede realizar? | `mostrar_informacion()` |

### Analogía

Una clase es como el plano de una casa. El plano define una estructura; las casas construidas a partir de él son objetos. Aunque compartan la estructura general, cada casa puede tener valores particulares, como color o dirección.

**Importante:** la clase no es el objeto. La clase define cómo crear objetos; cada objeto es una instancia concreta.


## 3. Primera clase en Python

En Python definimos una clase con la palabra clave `class`. Por convención, los nombres de las clases suelen escribirse en `PascalCase`: cada palabra empieza con mayúscula, como `Mascota` o `CuentaBancaria`.

Empecemos con una clase sin atributos ni métodos propios:


````python
class Saludo:
    pass

# Creamos un objeto (una instancia) de la clase Saludo
saludo_1 = Saludo()

print(type(saludo_1))
````

La expresión `Saludo()` crea una instancia. En este caso, `saludo_1` es una variable que referencia el objeto creado.

## 4. Atributos y `self`

Podemos asociar atributos a un objeto. Para entender la idea, primero observemos cómo se asignan atributos después de crear la instancia:


````python
class Estudiante:
    pass

estudiante_1 = Estudiante()
estudiante_1.nombre = "Laura"
estudiante_1.programa = "Física"

print(estudiante_1.nombre)
print(estudiante_1.programa)
````

Esto funciona, pero no es la forma más organizada de definir los datos que esperamos que tenga cada estudiante. Queremos que todos los objetos `Estudiante` se creen con una estructura consistente.

Para ello usamos el método especial `__init__`, que Python ejecuta automáticamente cuando creamos una instancia. En esta clase lo utilizaremos para inicializar los atributos.

```python
class Estudiante:
    def __init__(self, nombre, programa):
        self.nombre = nombre
        self.programa = programa
```

- `def` define una función; dentro de una clase, esta función es un método.
- `__init__` es un método especial de inicialización.
- `nombre` y `programa` son parámetros que recibe el método.
- `self` hace referencia a la instancia concreta con la que estamos trabajando.
- `self.nombre = nombre` guarda el valor recibido en el atributo `nombre` del objeto.

`self` debe aparecer como primer parámetro de los métodos de instancia. Al llamar al método a través de un objeto, Python pasa la instancia automáticamente; normalmente no escribimos `self` en la llamada.

### Ejemplo guiado 1: clase `Estudiante`


````python
class Estudiante:
    def __init__(self, nombre, programa):
        self.nombre = nombre
        self.programa = programa

estudiante_1 = Estudiante("Laura", "Física")
estudiante_2 = Estudiante("Mateo", "Ingeniería")

print(estudiante_1.nombre)
print(estudiante_1.programa)
print(estudiante_2.nombre)
print(estudiante_2.programa)
````

## 5. Métodos: comportamiento de los objetos

Un método es una función definida dentro de una clase. Puede consultar los atributos del objeto, realizar cálculos o modificar su estado.

### Ejemplo guiado 2: una cuenta bancaria sencilla

La clase tendrá:
- Atributos: `titular` y `saldo`.
- Método `mostrar_saldo()`: muestra el saldo actual.
- Método `depositar(cantidad)`: aumenta el saldo si la cantidad es positiva.
- Método `retirar(cantidad)`: disminuye el saldo si la cantidad es positiva y no supera el saldo disponible.

Este ejemplo es didáctico; no representa un sistema bancario real.


````python
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = saldo_inicial

    def mostrar_saldo(self):
        print(f"Titular: {self.titular}")
        print(f"Saldo actual: ${self.saldo:.2f}")

    def depositar(self, cantidad):
        if cantidad > 0:
            self.saldo += cantidad
            print(f"Depósito realizado: ${cantidad:.2f}")
        else:
            print("El depósito debe ser mayor que cero.")

    def retirar(self, cantidad):
        if cantidad <= 0:
            print("El retiro debe ser mayor que cero.")
        elif cantidad > self.saldo:
            print("Saldo insuficiente.")
        else:
            self.saldo -= cantidad
            print(f"Retiro realizado: ${cantidad:.2f}")

cuenta = CuentaBancaria("Andrea", 100)
cuenta.mostrar_saldo()
cuenta.depositar(50)
cuenta.retirar(30)
cuenta.mostrar_saldo()
````

### Analicemos el código

- `saldo_inicial=0` define un valor predeterminado. Si no se entrega un saldo, el objeto comienza con saldo cero.
- `self.saldo += cantidad` equivale a `self.saldo = self.saldo + cantidad`.
- Los métodos modifican `self.saldo`; por eso el nuevo saldo se conserva en el objeto.
- Las condiciones evitan aceptar depósitos o retiros no positivos y evitan retirar más dinero del disponible.
- `:.2f` muestra un número con dos cifras decimales.

## 6. Un método puede devolver un resultado

Un método no tiene que imprimir siempre. Puede usar `return` para devolver un valor que el programa pueda guardar o utilizar en un cálculo.

Observa la diferencia:
- `print(...)` muestra información en pantalla.
- `return ...` entrega un resultado a quien llamó al método.



````python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def calcular_area(self):
        return self.ancho * self.alto

    def calcular_perimetro(self):
        return 2 * (self.ancho + self.alto)

rectangulo_1 = Rectangulo(4, 3)

area = rectangulo_1.calcular_area()
perimetro = rectangulo_1.calcular_perimetro()

print("Área:", area)
print("Perímetro:", perimetro)
````

## 7. Errores frecuentes y buenas prácticas

### Error 1: olvidar `self` en la definición de un método

Incorrecto:
```python
class Persona:
    def saludar():
        print("Hola")
```

Al llamar `persona.saludar()`, Python pasa la instancia automáticamente. Como el método no recibe ese argumento, se produce un error. Para un método de instancia, escribe `def saludar(self):`.

### Error 2: olvidar `self.` al guardar un atributo

Incorrecto:
```python
def __init__(self, nombre):
    nombre = nombre
```

Esta asignación no crea el atributo del objeto. Lo correcto es `self.nombre = nombre`.

### Error 3: confundir la clase con una instancia

- `Estudiante` es el nombre de la clase.
- `Estudiante("Laura", "Física")` crea una instancia.
- `estudiante_1.nombre` consulta un atributo del objeto.

### Error 4: llamar un método sin paréntesis

`cuenta.mostrar_saldo` se refiere al método; `cuenta.mostrar_saldo()` lo ejecuta.

### Buenas prácticas iniciales

- Usa nombres de clase en `PascalCase`: `CuentaBancaria`.
- Usa nombres descriptivos para atributos y métodos: `saldo`, `calcular_area()`.
- Cada método debe tener una responsabilidad clara.
- Comprueba entradas cuando sea importante para el comportamiento esperado.
- No agregues clases por obligación: úsalas cuando ayuden a representar y organizar el problema.

## 10. Resumen de la clase

- **Clase:** define una estructura común para crear objetos.
- **Objeto:** instancia concreta de una clase.
- **Atributo:** dato asociado a un objeto.
- **Método:** función definida en una clase.
- **`__init__`:** inicializa un objeto al crearlo.
- **`self`:** referencia a la instancia actual dentro de los métodos de instancia.
- **`return`:** permite devolver un resultado para utilizarlo fuera del método.
