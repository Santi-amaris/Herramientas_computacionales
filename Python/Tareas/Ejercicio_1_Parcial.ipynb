```python
#Entrada de datos
nombre_est= input("Ingrese el nombre del estudiante: ",)
nota_corte_1= float(input("Ingrese la nota del primer corte: ",))
nota_corte_2= float(input("Ingrese la nota del segundo corte: ",))
nota_corte_3= float(input("Ingrese la nota del tercer corte: ",))

#Procesamiento de datos y declaración de variables
nota_def= (nota_corte_1*0.30) + (nota_corte_2*0.30) + (nota_corte_3*0.40)

if (4.5<=nota_def<=5.0):
    resul_academico= "Excelente"
elif (4.0<=nota_def<=4.49):
    resul_academico= "Muy buen desempeño"
elif (3.0<=nota_def<=3.99):
    resul_academico= "Aprobado"
elif (2.0<=nota_def<=2.99):
    resul_academico= "Reprobado"
elif (0.0<=nota_def<=1.99):
    resul_academico= "Reprobado con bajo desempeño"
elif(((nota_corte_1<0) or (nota_corte_1>5)) or ((nota_corte_2<0) or (nota_corte_2>5)) or ((nota_corte_3<0) or (nota_corte_3>5))):
        resul_academico= "Nulo"
        print("Una o más notas ingresadas no son válidas")
#resul_academico= "Nulo" es una bandera para que el sistema no imprima los datos a continuación, ya que en vez de meter todo en un gran if, preferí separarlos...

#Salida de datos
if (resul_academico!="Nulo"):
    print("Bienvenido al sistema. Los resultados son los siguientes: ")
    print("-----------------------------------------")
    print("Nombre del estudiante:",nombre_est)
    print("-----------------------------------------")
    print("La nota del primer corte es:",nota_corte_1)
    print("-----------------------------------------")
    print("La nota del segundo corte es:",nota_corte_2)
    print("-----------------------------------------")
    print("La nota del tercer corte es:",nota_corte_3)
    print("-----------------------------------------")
    print("La nota definitiva es la siguiente:", (round(nota_def, 2)))
    print("-----------------------------------------")
    print("El resultado académico es el siguiente:", resul_academico)
else:
    print("Error")
```
