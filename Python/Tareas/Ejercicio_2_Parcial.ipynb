#### $\color{Plum}{\underline{\textbf{Ejercicio 2 --- Sistema de facturación}}}$

```python
#Entrada de datos
nom_cliente= input("Ingrese el nombre del cliente",)
nom_produc= input("Ingrese el nombre del producto a pagar",)
precio_unit= float(input("Ingrese el precio unitario del producto a pagar"))
cant_uni= int(input("Ingrese la cantidad de unidades del producto a pagar",))
tarjeta_fidel= input("¿Posee tarjeta de fidelización? (S/N)",)

#Procesamiento de datos y declaración de variables
subtotal=(precio_unit*cant_uni)

if((precio_unit>0) and (cant_uni>0)):
    bandera= True
else:
    bandera=False #Nuevamente y para cumplir con los requisitos del problema, preferí añadir una bandera para que el sistema sepa si los resultados son útiles o no
    

if(subtotal< 100000):
    descu_precio=0
elif(100000<subtotal<199999):
    descu_precio=5
elif(subtotal>200000):
    descu_precio=10


if (tarjeta_fidel=="S" or tarjeta_fidel=="s"):
    tarjeta_fidel=True
    descu_tarjeta=5
elif (tarjeta_fidel=="N" or tarjeta_fidel=="n"):
    tarjeta_fidel=False
    descu_tarjeta=0
else:
    tarjeta_fidel=False
    descu_tarjeta=0

descu_tot= descu_precio + descu_tarjeta
valor_descu= ((subtotal)*(descu_tot/100))
tot_pagar= (subtotal - valor_descu)

if (bandera==True):
    print("Esta es su factura: ")
    print("--------------------------")
    print((nom_cliente))
    print("--------------------------")
    print("El producto comprado fue:",((nom_produc)))
    print("--------------------------")
    print("La cantidad de productos comprados fue:", cant_uni)
    print("--------------------------")
    print("El subtotal fue de:", subtotal)
    print("--------------------------")
    print("El porcentaje de descuento total fue:", descu_tot)
    print("--------------------------")
    print("EL valor del descuento fue:", valor_descu)
    print("--------------------------")
    print("El total a pagar fue:", tot_pagar)
else:
    print("Error")
```
